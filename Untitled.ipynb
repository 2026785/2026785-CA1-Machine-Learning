```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the dataset from Excel
df = pd.read_excel('Student_Depression_Dataset.xlsx')
```

```python
# Display first 5 rows and dataset
print("Dataset shape:", df.shape)
df.head()
```

```python
print("Missing values per column:\n", df.isnull().sum())
```

```python
df.describe()
```

```python
# Visualize the target variable distribution
plt.figure(figsize=(6, 4))
sns.countplot(x='Depression', data=df)
plt.title('Distribution of Depression (Target Variable)')
plt.xlabel('Depression')
plt.ylabel('Count')
plt.show()

print(df['Depression'].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder

# Convert categorical text columns into numbers
le = LabelEncoder()
for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col])
    
# Verify all columns are now numerical
df.head()
```

```python
# Check correlation of features with the target variable 'Depression'
correlations = df.corr()['Depression'].drop('Depression').sort_values()
print("Feature correlations with Depression:\n", correlations)

# Plot feature correlations
plt.figure(figsize=(8, 5))
correlations.plot(kind='barh', color='steelblue')
plt.title('Correlation of Features with Depression')
plt.xlabel('Correlation Coefficient')
plt.show()
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
# Define features (X) and target (y)
X = df.drop('Depression', axis=1)
y = df['Depression']
```

```python
# Three test splits: 5%, 15%, and 25%
splits = [0.05, 0.15, 0.25]

knn_test_scores = []
svm_test_scores = []

for s in splits:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=s, random_state=42)

    sc = StandardScaler()
    X_train_sc = sc.fit_transform(X_train)
    X_test_sc = sc.transform(X_test)

# Model 1: kNN
    knn = KNeighborsClassifier(n_neighbors=5)
    knn.fit(X_train_sc, y_train)
    knn_acc = accuracy_score(y_test, knn.predict(X_test_sc))
    knn_cv = cross_val_score(knn, X_train_sc, y_train, cv=5, scoring='accuracy').mean()
    knn_test_scores.append(knn_acc)

# Model 2: SVM
    svm = SVC(random_state=42)
    svm.fit(X_train_sc, y_train)
    svm_acc = accuracy_score(y_test, svm.predict(X_test_sc))
    svm_cv = cross_val_score(svm, X_train_sc, y_train, cv=5, scoring='accuracy').mean()
    svm_test_scores.append(svm_acc)

    print(f"--- Split test_size = {int(s*100)}% (Train = {int((1-s)*100)}%) ---")
    print(f"kNN -> Test Accuracy: {knn_acc:.4f} | 5-Fold CV Accuracy: {knn_cv:.4f}")
    print(f"SVM -> Test Accuracy: {svm_acc:.4f} | 5-Fold CV Accuracy: {svm_cv:.4f}\n")
```

```python
# Visualize accuracy comparison across splits
split_labels = ['5% Test', '15% Test', '25% Test']
x = np.arange(len(split_labels))
width = 0.35

plt.figure(figsize=(7, 4))
plt.bar(x - width/2, knn_test_scores, width, label='kNN')
plt.bar(x + width/2, svm_test_scores, width, label='SVM')
plt.xticks(x, split_labels)
plt.ylabel('Test Accuracy')
plt.ylim(0.80, 1.0)
plt.title('kNN vs SVM Accuracy across Splits')
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

# Use the 25% split (75% training, 25% testing)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

```python
# 1. GridSearchCV for kNN
param_grid_knn = {
    'n_neighbors': [3, 5, 7, 9],
    'metric': ['euclidean', 'manhattan']
}
grid_knn = GridSearchCV(KNeighborsClassifier(), param_grid_knn, cv=5)
grid_knn.fit(X_train_sc, y_train)

print("Best kNN Parameters:", grid_knn.best_params_)
print("Best kNN CV Accuracy:", round(grid_knn.best_score_, 4))
knn_best_pred = grid_knn.predict(X_test_sc)
print("Tuned kNN Test Accuracy:", round(accuracy_score(y_test, knn_best_pred), 4))
```

```python
# 2. GridSearchCV for SVM
param_grid_svm = {
    'C': [0.1, 1, 10],
    'gamma': [0.1, 0.01],
    'kernel': ['rbf']
}
grid_svm = GridSearchCV(SVC(), param_grid_svm, cv=5)
grid_svm.fit(X_train_sc, y_train)

print("\nBest SVM Parameters:", grid_svm.best_params_)
print("Best SVM CV Accuracy:", round(grid_svm.best_score_, 4))
svm_best_pred = grid_svm.predict(X_test_sc)
print("Tuned SVM Test Accuracy:", round(accuracy_score(y_test, svm_best_pred), 4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

# Check Training vs Test accuracy for Overfitting/Underfitting analysis
print("Tuned SVM Train Accuracy:", round(grid_svm.score(X_train_sc, y_train), 4))
print("Tuned SVM Test Accuracy: ", round(accuracy_score(y_test, svm_best_pred), 4))

# Classification Report for Best Model (SVM)
print("\nClassification Report (Tuned SVM):\n")
print(classification_report(y_test, svm_best_pred))

# Confusion Matrix Heatmap
cm = confusion_matrix(y_test, svm_best_pred)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix - Tuned SVM')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
